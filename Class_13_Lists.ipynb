## ক্লাস ১৩: Lists

### 🎯 আজকের লক্ষ্য
- List তৈরি ও ইনডেক্সিং
- লিস্টে আইটেম যোগ (`append`, `insert`) ও বাদ দেওয়া (`remove`, `pop`)
- সর্টিং ও ডুপ্লিকেট অনুলিপি (`copy`)
- লুপ দিয়ে লিস্টের ডেটা অ্যাক্সেস করা

```python
# ১. লিস্ট তৈরি ও পরিবর্তন
fruits = ["আম", "কাঁঠাল", "লিচু"]
print("লিস্ট:", fruits)

# যোগ করা
fruits.append("জাম")             # শেষে যোগ
fruits.insert(1, "কলা")          # নির্দিষ্ট পজিশনে যোগ
print("পরিবর্তিত লিস্ট:", fruits)

# বাদ দেওয়া
fruits.remove("কাঁঠাল")          # আইটেম সরাসরি বাদ
popped = fruits.pop()            # শেষেরটি বের করা
print("রিমুভ করার পর:", fruits)
print("বের করা আইটেম:", popped)
```

```python
# ২. সর্টিং ও কপি
nums = [5, 1, 9, 3, 2]
nums.sort()
print("Sorted:", nums)

# কপি
nums_copy = nums.copy()
print("কপি করা লিস্ট:", nums_copy)
```

```python
# ৩. লুপ দিয়ে লিস্ট প্রিন্ট করা
students = ["রাফি", "সানা", "করিম"]
for i, student in enumerate(students, 1):
    print(f"{i} নম্বর ছাত্র: {student}")
```
