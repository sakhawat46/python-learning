### 🎯 আজকের লক্ষ্য
- স্ট্রিং নিয়ে চমৎকার প্র্যাকটিক্যাল সমস্যা সমাধান
- Palindrome Check (প্যালিন্ড্রোম যাচাই)
- Word Counter (শব্দ গণনাকারী)
- Password Validation (পাসওয়ার্ডের সঠিকতা যাচাই)

```python
# ১. Palindrome Check (উভয় দিক থেকে পড়লে সমান শব্দ, যেমন: madam, radar)
word = input("একটি শব্দ লিখো: ").lower().strip()
reversed_word = word[::-1]

if word == reversed_word:
    print(f"✅ '{word}' একটি প্যালিন্ড্রোম শব্দ!")
else:
    print(f"❌ '{word}' প্যালিন্ড্রোম নয়।")
```

```python
# ২. Word Counter (শব্দ গণনাকারী)
sentence = input("একটি বাক্য লিখো: ")
words = sentence.split()
print("বাক্যের মোট শব্দ সংখ্যা হলো:", len(words))
```

```python
# ৩. Password Validation (পাসওয়ার্ড চেক)
# শর্ত: পাসওয়ার্ড কমপক্ষে ৮ অক্ষরের হতে হবে এবং তাতে সংখ্যা ও টেক্সট দুইটাই থাকতে হবে
password = input("একটি নতুন পাসওয়ার্ড দাও: ")

has_num = False
has_char = False

for char in password:
    if char.isdigit():
        has_num = True
    if char.isalpha():
        has_char = True

if len(password) >= 8 and has_num and has_char:
    print("✅ পাসওয়ার্ডটি গ্রহণযোগ্য ও শক্তিশালী!")
else:
    print("❌ পাসওয়ার্ড দুর্বল! কমপক্ষে ৮ ক্যারেক্টার ও সংখ্যা+অক্ষর মিলিয়ে দিন।")
```
