- ব্যাঙ্ক অ্যাকাউন্ট হায়ারার্কি সিস্টেম (Savings vs Current Account) তৈরি
- OOP-র চারটি মূল স্তম্ভ (Encapsulation, Inheritance, Polymorphism, Abstraction) এর পূর্ণাঙ্গ সমন্বয়

```python
# ১. বাস্তবমুখী প্রজেক্ট: Library Management System
# ক্লাসসমূহ: Book, Member, Library

class Book:
    def __init__(self, book_id, title, author):
        self.book_id = book_id
        self.title = title
        self.author = author
        self.is_borrowed = False

    def __str__(self):
        status = "❌ ধার নেওয়া হয়েছে" if self.is_borrowed else "✅ পাওয়া যাবে"
        return f"[{self.book_id}] '{self.title}' - {self.author} ({status})"


class Member:
    def __init__(self, member_id, name):
        self.member_id = member_id
        self.name = name
        self.borrowed_books = []

    def borrow_book(self, book):
        if book.is_borrowed:
            print(f"দুঃখিত {self.name}, '{book.title}' বইটি ইতিমধ্যে অন্য কেউ নিয়েছে।")
        else:
            book.is_borrowed = True
            self.borrowed_books.append(book)
            print(f"🎉 {self.name} সফলভাবে '{book.title}' বইটি ধার নিয়েছেন।")

    def return_book(self, book):
        if book in self.borrowed_books:
            book.is_borrowed = False
            self.borrowed_books.remove(book)
            print(f"🔄 {self.name} '{book.title}' বইটি সফলভাবে ফেরত দিয়েছেন।")
        else:
            print(f"⚠️ এই বইটি {self.name}-এর কাছে নেই!")


class Library:
    def __init__(self, name):
        self.name = name
        self.books = []

    def add_book(self, book):
        self.books.append(book)
        print(f"বই যুক্ত হলো: {book.title}")

    def show_available_books(self):
        print(f"\n📚 {self.name}-এর বর্তমান বইয়ের তালিকা:")
        for b in self.books:
            print(b)


# ডেমো রান
my_lib = Library("কেন্দ্রীয় পাবলিক লাইব্রেরি")
b1 = Book("B101", "পাইথন প্রোগ্রামিং", "তামিম শাহরিয়ার সুবিন")
b2 = Book("B102", "হিমু", "হুমায়ূন আহমেদ")
my_lib.add_book(b1)
my_lib.add_book(b2)

member1 = Member("M01", "রাফি")
my_lib.show_available_books()

member1.borrow_book(b1)
my_lib.show_available_books()

member1.return_book(b1)
my_lib.show_available_books()
```

```python
# ২. বাস্তবমুখী প্রজেক্ট: Bank Account Hierarchy System
# এবস্ট্রাকশন, ইনহেরিটেন্স ও এনক্যাপসুলেশন এর বাস্তব প্রয়োগ

from abc import ABC, abstractmethod

class BaseAccount(ABC):
    def __init__(self, acc_no, holder_name, balance=0):
        self.acc_no = acc_no
        self.holder_name = holder_name
        self._balance = balance

    @abstractmethod
    def withdraw(self, amount):
        pass

    def deposit(self, amount):
        if amount > 0:
            self._balance += amount
            print(f"✅ {amount} টাকা জমা হলো। নতুন ব্যালেন্স: {self._balance}")
        else:
            print("⚠️ অবৈধ ডিপোজিট পরিমাণ!")

    def get_balance(self):
        return self._balance


class SavingsAccount(BaseAccount):
    interest_rate = 0.05  # ৫% বার্ষিক সুদ

    def withdraw(self, amount):
        # সেভিংস একাউন্টে মিনিমাম ৫০০ টাকা ব্যালেন্স থাকা বাধ্যতামূলক
        if self._balance - amount >= 500:
            self._balance -= amount
            print(f"💸 সেভিংস অ্যাকাউন্ট থেকে {amount} টাকা উত্তোলন সফল। ব্যালেন্স: {self._balance}")
        else:
            print("❌ উত্তোলন ব্যর্থ! সেভিংস অ্যাকাউন্টে ন্যূনতম ৫০০ টাকা ব্যালেন্স থাকতে হবে।")

    def apply_interest(self):
        interest = self._balance * self.interest_rate
        self._balance += interest
        print(f"📈 {interest} টাকা সুদ যোগ হয়েছে। মোট ব্যালেন্স: {self._balance}")


class CurrentAccount(BaseAccount):
    def __init__(self, acc_no, holder_name, balance=0, overdraft_limit=2000):
        super().__init__(acc_no, holder_name, balance)
        self.overdraft_limit = overdraft_limit  # ব্যাংক ঋণ/ওভারড্রাফট সুবিধা

    def withdraw(self, amount):
        if self._balance + self.overdraft_limit >= amount:
            self._balance -= amount
            print(f"💸 কারেন্ট অ্যাকাউন্ট থেকে {amount} টাকা উত্তোলন সফল। অবশিষ্ট ব্যালেন্স: {self._balance}")
        else:
            print("❌ ওভারড্রাফট লিমিট অতিক্রম করেছে! উত্তোলন সম্ভব নয়।")


# ডেমো
savings = SavingsAccount("SA-101", "সাকিব", 2000)
savings.withdraw(1600)  # ব্যর্থ হবে কারণ ৫০০ টাকার কম থাকবে
savings.withdraw(1200)  # সফল
savings.apply_interest()

current = CurrentAccount("CA-201", "করিম এন্টারপ্রাইজ", 1000, overdraft_limit=3000)
current.withdraw(2500)  # সফল (ব্যালেন্স হবে -১৫০০, যা লিমিটের মধ্যে)
```
