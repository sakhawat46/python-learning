### 🐍 Python OOP কী?
**OOP = Object-Oriented Programming**

সহজ ভাষায়: কোনো বাস্তব জিনিসকে (**Object**) প্রোগ্রামিংয়ের মধ্যে মডেল বা রিপ্রেজেন্ট করার পদ্ধতিই হলো OOP।

যেমন, বাস্তবে একজন **Student**-এর কথা চিন্তা করো:
- তার নাম আছে, বয়স আছে, ইমেইল আছে।
- সে পড়াশোনা করে, সে পরীক্ষায় অংশ নেয়।

প্রোগ্রামিংয়ে আমরা এগুলোকে আলাদা আলাদা চলকে না রেখে, একসাথে একটি **Student Object** হিসেবে সুন্দরভাবে গুছিয়ে রাখতে পারি।

### ১. Class কী?
**Class হলো একটি Blueprint বা নকশা।**

যেমন বাড়ি বানানোর আগে বাড়ির একটি ডিজাইন বা নকশা থাকে, সেই নকশা দেখে যত ইচ্ছা বাড়ি বানানো যায়। তেমনি ক্লাস হলো অবজেক্ট তৈরির নকশা।

```
Class = Student (Blueprint)
        ↓
 ┌─────────────┐
 │ student1    │
 │ student2    │  (বাস্তব Objects)
 │ student3    │
 └─────────────┘
```

**মনে রাখো:**
- **Class** $\rightarrow$ Blueprint (নকশা)
- **Object** $\rightarrow$ Blueprint থেকে তৈরি বাস্তব জিনিস

````python
# ১. ব্লুপ্রিন্ট (Class) ও অবজেক্ট তৈরি
class StudentBasic:
    pass  # খালি ক্লাস

# এই ব্লুপ্রিন্ট থেকে একাধিক অবজেক্ট তৈরি
student1 = StudentBasic()
student2 = StudentBasic()
student3 = StudentBasic()

print("student1 অবজেক্ট:", student1)
print("student2 অবজেক্ট:", student2)
````

---
### ২. Attribute, `__init__()` এবং `self` কী?

- **Attribute (বৈশিষ্ট্য):** Object-এর যে data বা property থাকে সেটাকে attribute বলে (যেমন: `name`, `age`, `email`)।
- **`__init__()` (Constructor):** এটি একটি বিশেষ মেথড যা অবজেক্ট তৈরি করার সময় automatically execute হয় এবং অবজেক্টের প্রাথমিক ডেটা সেট করে।
- **`self` কী?:** শুরুতে `self` একটু কনফিউজিং লাগতে পারে। সহজ কথায় — `self` মানে হলো **বর্তমানে যে object নিয়ে কাজ করছি**।
  - যখন লিখি `student1 = Student("Rahim", 20)`, তখন `self = student1`, অর্থাৎ `self.name` মানে `student1.name`।
  - যখন লিখি `student2 = Student("Karim", 22)`, তখন `self = student2`, অর্থাৎ `self.name` মানে `student2.name`।

````python
# ২. Attribute, __init__ এবং self এর ব্যবহার
class Student:
    def __init__(self, name, age):
        self.name = name  # Attribute
        self.age = age    # Attribute

# অবজেক্ট তৈরি (স্বয়ংক্রিয়ভাবে __init__ কল হবে)
student1 = Student("Rahim", 20)
student2 = Student("Karim", 22)

print(student1.name)
print(student1.age)

print(student2.name)
print(student2.age)
````

---
### ৩. Method কী?
Class-এর ভিতরের function-কে সাধারণত **Method** বলা হয়। মেথড মূলত অবজেক্টের বিভিন্ন অ্যাকশন বা আচরণ পরিচালনা করে।

````python
# ৩. Method সহ পূর্ণাঙ্গ Student ক্লাস
class StudentFull:
    def __init__(self, name, age, marks):
        self.name = name
        self.age = age
        self.marks = marks

    def study(self):
        print(self.name, "is studying")

    def show_result(self):
        print(self.name, "got", self.marks, "marks")

# অবজেক্ট তৈরি ও ডেটা অ্যাক্সেস
student1 = StudentFull("Rahim", 20, 85)
student2 = StudentFull("Karim", 21, 90)

print(student1.name)
print(student1.age)
print(student1.marks)

# মেথড কল
student1.study()
student1.show_result()
````

### ৪. অবজেক্ট প্রোপার্টি পরিবর্তন ও মুছে ফেলা (Modify & Delete)
অবজেক্টের ডেটা সরাসরি পরিবর্তন করা যায়, এবং `del` কি-ওয়ার্ড দিয়ে অপ্রয়োজনীয় প্রোপার্টি ডিলিট করা যায়।

````python
# ৪. অবজেক্ট প্রোপার্টি পরিবর্তন ও ডিলিট করা
student1.marks = 95  # প্রোপার্টি পরিবর্তন
print("সংশোধিত নম্বর:", student1.marks)

# ডিলিট প্রোপার্টি
del student1.marks
print("marks প্রোপার্টি মুছে ফেলা হয়েছে।")
# print(student1.marks)  # এটি AttributeError দিবে কারণ marks ডিলিট করা হয়েছে
````

### ৫. Car ক্লাসের বাস্তব উদাহরণ ও সামারি টেবিল

````python
# ৫. Car ক্লাস উদাহরণ
class Car:
    def __init__(self, brand, color):
        self.brand = brand
        self.color = color

    def drive(self):
        print(self.brand, "is driving")

car1 = Car("Toyota", "White")
car2 = Car("BMW", "Black")

car1.drive()
car2.drive()
````

| বিষয় | কোডে উদাহরণ | অর্থ ও কাজ |
| :--- | :--- | :--- |
| **Class** | `Car` | ব্লুপ্রিন্ট বা নকশা |
| **Object** | `car1`, `car2` | ব্লুপ্রিন্ট থেকে তৈরি বাস্তব অবজেক্ট |
| **Constructor** | `__init__()` | অবজেক্ট তৈরির সময় প্রাথমিক মান নির্ধারণকারী |
| **Attribute** | `brand`, `color` | অবজেক্টের ডেটা বা বৈশিষ্ট্য |
| **Method** | `drive()` | অবজেক্টের কাজ বা অ্যাকশন |
| **self** | `self` | বর্তমান অবজেক্টের নিজস্ব রেফারেন্স |

### ৬. Food Delivery App এনালজি
একটি Food Delivery App এর কথা চিন্তা করলে একাধিক অবজেক্ট দেখা যায়:
- `User` (ব্যবহারকারী)
- `Restaurant` (রেস্তোরাঁ)
- `Food` (খাবার)
- `Order` (অর্ডার)
- `Payment` (পেমেন্ট)
- `Delivery` (ডেলিভারি)

যেমন, `User` ক্লাসের প্রতিটি অবজেক্টের আলাদা নাম, ইমেইল এবং লগইন মেথড থাকবে:

````python
# ৬. Food Delivery App - User ক্লাস
class User:
    def __init__(self, name, email):
        self.name = name
        self.email = email

    def login(self):
        print(self.name, "logged in")

user1 = User("Rahim", "rahim@gmail.com")
user2 = User("Karim", "karim@gmail.com")

user1.login()
user2.login()
````

### 🌟 ৭. OOP-এর ৪টা Main Concept (The 4 Pillars)

```
             OOP
              │
      ┌───────┼────────┬────────┐
      ↓       ↓        ↓        ↓
Encapsulation Inheritance Polymorphism Abstraction
```

1. **Encapsulation (এনক্যাপসুলেশন):** Data এবং methods-কে একসাথে রাখা।
2. **Inheritance (উত্তরাধিকার):** একটি class অন্য class-এর property/method ব্যবহার করা।
3. **Polymorphism (বহুরূপতা):** একই method name, কিন্তু different behavior।
4. **Abstraction (বিমূর্তকরণ):** অপ্রয়োজনীয় implementation details লুকিয়ে শুধু প্রয়োজনীয় interface দেখানো (যেমন ATM মেশিন)।

> 📌 **নোট:** এই ৪টি মূল বিষয় নিয়ে আমাদের পরবর্তী ক্লাসসমূহ সাজানো হয়েছে: **ক্লাস ২৫** (Inheritance ও Encapsulation), **ক্লাস ২৬** (Class/Static Methods ও @property), **ক্লাস ২৭** (Polymorphism ও Abstraction) এবং **ক্লাস ২৮** (বাস্তবমুখী OOP প্রজেক্ট প্র্যাকটিস)।

---
### 🧠 OOP-এর পুরো বিষয়টা এক নজরে (Cheatsheet)
- **Class** $\rightarrow$ Blueprint (নকশা)
- **Object** $\rightarrow$ Class থেকে তৈরি বাস্তব instance
- **Attribute** $\rightarrow$ Object-এর data
- **Method** $\rightarrow$ Object-এর কাজ
- **Inheritance** $\rightarrow$ এক class অন্য class-এর feature নেয়
- **Polymorphism** $\rightarrow$ একই method $\rightarrow$ different behavior
- **Encapsulation** $\rightarrow$ Data + methods একসাথে রাখা
- **Abstraction** $\rightarrow$ জটিল implementation লুকিয়ে প্রয়োজনীয় অংশ দেখানো
