## ক্লাস ২১: File Handling

### 🎯 আজকের লক্ষ্য
- ফাইলে ডেটা লেখা (`write`) এবং ফাইল রিড করা (`read`)
- ফাইলে ডেটা সংযোজন করা (`append`)
- ফাইল তৈরি ও ডিলিট করা (`os` মডিউল ব্যবহার করে)

```python
# ১. ফাইলে লেখা ও পড়া
# 'w' মোড ফাইল তৈরি করে এবং নতুন করে লেখে। utf-8 বাংলা সাপোর্টের জন্য উপযোগী।
with open("student_notes.txt", "w", encoding="utf-8") as f:
    f.write("পাইথন ফাইল হ্যান্ডলিং শেখা সহজ।\n")
    f.write("আমরা ফাইলে ডেটা সেভ করছি।\n")
print("ফাইল সফলভাবে তৈরি হয়েছে।")
```

```python
# ২. ফাইল রিড করা
with open("student_notes.txt", "r", encoding="utf-8") as f:
    content = f.read()
print("ফাইলের ভেতরের তথ্য:")
print(content)
```

```python
# ৩. ফাইলে সংযোজন (append) করা
with open("student_notes.txt", "a", encoding="utf-8") as f:
    f.write("এটি একটি নতুন সংযোজিত লাইন।\n")

# পুনরায় পড়া
with open("student_notes.txt", "r", encoding="utf-8") as f:
    print(f.read())
```

```python
# ৪. ফাইল নিরাপদে মুছে ফেলা
import os
temp_file = "temp.txt"
with open(temp_file, "w") as f:
    f.write("অস্থায়ী")

if os.path.exists(temp_file):
    os.remove(temp_file)
    print(f"✅ {temp_file} মুছে ফেলা হয়েছে।")
```
