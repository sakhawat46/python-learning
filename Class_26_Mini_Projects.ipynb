### 🎯 আজকের লক্ষ্য
- এখন পর্যন্ত শেখা সবকিছু মিলিয়ে বাস্তবমুখী প্রজেক্ট তৈরি
- ১. ক্যালকুলেটর অ্যাপ্লিকেশন (Console Calculator)
- ২. ইন্টারেক্টিভ কুইজ অ্যাপ্লিকেশন (Quiz Application)
- ৩. স্টুডেন্ট রেজাল্ট ম্যানেজমেন্ট টেমপ্লেট

```python
# ১. Console Calculator
def show_menu():
    print("\n--- ক্যালকুলেটর মেনু ---")
    print("1. যোগ  2. বিয়োগ  3. গুণ  4. ভাগ  5. বন্ধ করুন")

while True:
    show_menu()
    choice = input("পছন্দ করো (১-৫): ")
    
    if choice == '5':
        print("বিদায়!")
        break
        
    if choice in ['1', '2', '3', '4']:
        num1 = float(input("প্রথম সংখ্যা: "))
        num2 = float(input("দ্বিতীয় সংখ্যা: "))
        
        if choice == '1': print("ফলাফল:", num1 + num2)
        elif choice == '2': print("ফলাফল:", num1 - num2)
        elif choice == '3': print("ফলাফল:", num1 * num2)
        elif choice == '4':
            if num2 == 0: print("ভুল! শূন্য দিয়ে ভাগ করা যাবে না।")
            else: print("ফলাফল:", num1 / num2)
    else:
        print("অবৈধ পছন্দ।")
```

```python
# ২. Interactive Quiz Application
questions = [
    {
        "q": "পাইথন ভাষার আবিষ্কারক কে?",
        "options": ["A. Guido van Rossum", "B. Dennis Ritchie", "C. Bjarne Stroustrup"],
        "ans": "A"
    },
    {
        "q": "কোনটি পাইথনে মিউটেবল (পরিবর্তনযোগ্য) কালেকশন?",
        "options": ["A. Tuple", "B. List", "C. String"],
        "ans": "B"
    }
]

score = 0
print("🎮 কুইজ গেম শুরু হচ্ছে!\n")

for i, q in enumerate(questions, 1):
    print(f"প্রশ্ন {i}: {q['q']}")
    for opt in q['options']:
        print(opt)
    ans = input("তোমার উত্তর (A, B, C): ").upper().strip()
    if ans == q['ans']:
        print("সঠিক! ✅\n")
        score += 1
    else:
        print(f"ভুল! সঠিক উত্তরটি ছিল: {q['ans']}\n")

print(f"কুইজ শেষ! তোমার মোট প্রাপ্ত স্কোর: {score}/{len(questions)}")
```
