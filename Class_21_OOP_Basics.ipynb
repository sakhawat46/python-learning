## ক্লাস ২১: OOP Basics

### 🎯 আজকের লক্ষ্য
- Class এবং Object এর বাস্তবমুখী ধারণা
- Constructor (`__init__`) এবং ইনস্ট্যান্স ভ্যারিয়েবল
- অবজেক্টের প্রোপার্টি পরিবর্তন ও ডিলিট করা

```python
# ১. Class ও Object তৈরি
class Student:
    # কনস্ট্রাকটর (Constructor)
    def __init__(self, name, roll, gpa):
        self.name = name          # ইনস্ট্যান্স ভ্যারিয়েবল
        self.roll = roll
        self.gpa = gpa
    
    # ক্লাসের মেথড
    def display_info(self):
        print(f"ছাত্র: {self.name} | রোল: {self.roll} | GPA: {self.gpa}")

# অবজেক্ট তৈরি
s1 = Student("রাফি", 101, 4.85)
s2 = Student("সানা", 102, 5.00)

s1.display_info()
s2.display_info()
```

```python
# ২. অবজেক্ট প্রোপার্টি পরিবর্তন ও মুছে ফেলা (W3Schools স্পেশাল)
s1.gpa = 5.00  # প্রোপার্টি পরিবর্তন
print("পরিবর্তিত GPA:", s1.gpa)

# ডিলিট প্রোপার্টি
del s1.gpa
# print(s1.gpa)  # এটি AttributeError দিবে কারণ gpa ডিলিট করা হয়েছে
```
