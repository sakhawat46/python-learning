## ক্লাস ৪: If Statement

### 🎯 আজকের লক্ষ্য
- `if` এবং `if-else` এর গঠন
- Nested `if` (শর্তের ভেতর শর্ত)
- Real-life উদাহরণের মাধ্যমে সিদ্ধান্ত নেওয়া শেখা

```python
# ১. সাধারণ if-else গঠন
age = int(input("তোমার বয়স লিখো: "))

if age >= 18:
    print("তুমি ভোট দেওয়ার যোগ্য! ✅")
else:
    print("দুঃখিত, ভোট দিতে তোমার বয়স কম। ❌")
```

```python
# ২. Nested if (শর্তের ভেতর শর্ত)
number = int(input("একটি সংখ্যা দাও: "))

if number >= 0:
    if number == 0:
        print("সংখ্যাটি শূন্য (Zero)")
    else:
        print("সংখ্যাটি পজিটিভ (Positive)")
else:
    print("সংখ্যাটি নেগেটিভ (Negative)")
```
