### 🎯 আজকের লক্ষ্য
- কন্ডিশনাল স্টেটমেন্টের জটিল সমস্যা সমাধান
- HSC পরীক্ষার উপযোগী প্রোগ্রামিং সমস্যা সমাধান

```python
# সমস্যা ১: তিনটি সংখ্যার মধ্যে বড় সংখ্যাটি খুঁজে বের করো
num1 = float(input("প্রথম সংখ্যা: "))
num2 = float(input("দ্বিতীয় সংখ্যা: "))
num3 = float(input("তৃতীয় সংখ্যা: "))

if num1 >= num2 and num1 >= num3:
    largest = num1
elif num2 >= num1 and num2 >= num3:
    largest = num2
else:
    largest = num3

print(f"সবচেয়ে বড় সংখ্যাটি হলো: {largest}")
```

```python
# সমস্যা ২: দ্বিঘাত সমীকরণের মূল নির্ণয় (Quadratic Equation Roots) - HSC Special
# ax² + bx + c = 0 সমীকরণের জন্য d = b² - 4ac
import math

a = float(input("a এর মান: "))
b = float(input("b এর মান: "))
c = float(input("c এর মান: "))

d = (b**2) - (4*a*c)

if d > 0:
    root1 = (-b + math.sqrt(d)) / (2*a)
    root2 = (-b - math.sqrt(d)) / (2*a)
    print(f"সমীকরণের মূল দুটি বাস্তব ও অসমান: {root1:.2f} এবং {root2:.2f}")
elif d == 0:
    root = -b / (2*a)
    print(f"সমীকরণের মূল দুটি বাস্তব ও সমান: {root:.2f}")
else:
    print("সমীকরণের কোনো বাস্তব মূল নেই (জটিল মূল)।")
```
