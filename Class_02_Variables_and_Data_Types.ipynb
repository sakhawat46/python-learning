## ক্লাস ২: Variables & Data Types

### 🎯 আজকের লক্ষ্য
- Variable (চলক) এবং এর নামকরণের নিয়মাবলী
- মূল ৪টি Data Types (int, float, string, boolean) চেনা
- Input ও Output নেওয়া
- Type Conversion (টাইপ কাস্টিং) শেখা

### 1. Variables কী?
ভ্যারিয়েবল হলো তথ্য বা ডেটা রাখার একটি পাত্র বা বাক্স।

### 2. Variable নামকরণের নিয়মাবলী (W3Schools):
- নাম অবশ্যই বর্ণ (A-Z, a-z) অথবা আন্ডারস্কোর (_) দিয়ে শুরু হবে।
- নাম কোনো সংখ্যা দিয়ে শুরু করা যাবে না (যেমন: `1name` অবৈধ)।
- নামে কোনো স্পেস বা বিশেষ চিহ্ন থাকতে পারবে না।
- পাইথনের রিজার্ভড কিওয়ার্ড (যেমন: `if`, `for`, `print`) চলকের নাম হিসেবে ব্যবহার করা যাবে না।

```python
# ভ্যারিয়েবল ও ডেটা টাইপ ডিক্লেয়ারেশন
age = 17                  # int (পূর্ণ সংখ্যা)
gpa = 4.85                # float (দশমিক সংখ্যা)
name = "Mina"             # string (লেখা/টেক্সট)
is_hsc_student = True     # boolean (সত্য/মিথ্যা)

print(type(age))
print(type(gpa))
print(type(name))
print(type(is_hsc_student))
```

```python
# একসাথে একাধিক ভ্যারিয়েবল অ্যাসাইন ও আনপ্যাকিং
x, y, z = "Orange", "Banana", "Cherry"
print(x, y, z)

# Unpacking a list
fruits = ["আম", "কলা", "লিচু"]
a, b, c = fruits
print(a, b, c)
```

```python
# Input ও Output এবং Type Conversion
user_name = input("তোমার নাম কী? ")
birth_year = int(input("তোমার জন্ম সাল কত? "))  # কাস্টিং করে int করা হলো
age = 2026 - birth_year

print(f"স্বাগতম, {user_name}! ২০২৬ সালে তোমার বয়স হবে {age} বছর।")
```
