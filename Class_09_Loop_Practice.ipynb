### 🎯 আজকের লক্ষ্য
- লুপ দিয়ে ধারা ও গাণিতিক সমস্যার সমাধান করা
- নামতা, ফ্যাক্টোরিয়াল, মৌলিক সংখ্যা এবং ফিবোনাচ্চি ধারা তৈরি

```python
# ১. Multiplication Table (নামতা)
num = int(input("কোন সংখ্যার নামতা দেখতে চাও? "))
for i in range(1, 11):
    print(f"{num} x {i} = {num * i}")
```

```python
# ২. Factorial (ফ্যাক্টোরিয়াল: n!)
n = int(input("কোন সংখ্যার ফ্যাক্টোরিয়াল বের করতে চাও? "))
factorial = 1
for i in range(1, n + 1):
    factorial *= i
print(f"{n} এর ফ্যাক্টোরিয়াল হলো: {factorial}")
```

```python
# ৩. Prime Number Check (মৌলিক সংখ্যা যাচাই)
p = int(input("যাচাই করতে একটি সংখ্যা দাও: "))
is_prime = True

if p <= 1:
    is_prime = False
else:
    for i in range(2, int(p**0.5) + 1):
        if p % i == 0:
            is_prime = False
            break

if is_prime:
    print(f"{p} একটি মৌলিক সংখ্যা (Prime Number) ✅")
else:
    print(f"{p} মৌলিক সংখ্যা নয় ❌")
```

```python
# ৪. Fibonacci Series (ফিবোনাচ্চি ধারা: ০, ১, ১, ২, ৩, ৫, ৮...)
terms = int(input("কতটি ফিবোনাচ্চি সংখ্যা দেখতে চাও? "))
n1, n2 = 0, 1
count = 0

if terms <= 0:
    print("দয়া করে পজিটিভ সংখ্যা দাও।")
elif terms == 1:
    print("ফিবোনাচ্চি ধারা:", n1)
else:
    print("ফিবোনাচ্চি ধারা:", end=" ")
    while count < terms:
        print(n1, end=" ")
        nth = n1 + n2
        n1 = n2
        n2 = nth
        count += 1
print()
```
