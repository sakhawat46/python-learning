## ক্লাস ৭: For Loop

### 🎯 আজকের লক্ষ্য
- `for` লুপের গঠন ও ব্যবহার
- `range()` ফাংশনের বিভিন্ন মাত্রা
- প্যাটার্ন প্রিন্টিং (Pattern Printing) শেখা

````python
# ১. সাধারণ for loop এবং range()
# range(start, stop, step)
print("১ থেকে ৫ পর্যন্ত প্রিন্ট:")
for i in range(1, 6):
    print(i, end=" ")
print()

print("২ এর ব্যবধানে ০ থেকে ১০:")
for i in range(0, 11, 2):
    print(i, end=" ")
print()
````

````python
# ২. loop এর ভেতর else ব্যবহার (W3Schools স্পেশাল)
for x in range(3):
    print(x)
else:
    print("লুপের কাজ সফলভাবে শেষ হয়েছে!")
````

````python
# ৩. প্যাটার্ন প্রিন্টিং (Nested for loop)
rows = 5
for i in range(1, rows + 1):
    for j in range(i):
        print("*", end=" ")
    print()
````
