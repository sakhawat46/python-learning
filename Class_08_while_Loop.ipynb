# মডিউল ৩: Loops
## ক্লাস ৮: While Loop

### 🎯 আজকের লক্ষ্য
- `while` লুপের গঠন ও শর্ত
- Infinite Loop (অসীম লুপ) এবং কীভাবে তা পরিহার করা যায়
- `break` এবং `continue` স্টেটমেন্টের সঠিক ব্যবহার

```python
# ১. সাধারণ while loop
count = 1
while count <= 5:
    print("গণনা:", count)
    count += 1
```

```python
# ২. break এবং continue এর ব্যবহার
print("--- break এর খেলা ---")
i = 1
while i < 10:
    if i == 5:
        break   # ৫ এ পৌঁছালে লুপ থেকে বের হয়ে যাবে
    print(i, end=" ")
    i += 1
print()

print("--- continue এর খেলা ---")
i = 0
while i < 6:
    i += 1
    if i == 3:
        continue  # ৩ কে বাদ দিয়ে পরের ইটারেশনে চলে যাবে
    print(i, end=" ")
```

```python
# ৩. Infinite loop এড়াতে সতর্ক হওয়া
# নিচের লুপটি অসীম পর্যন্ত চলত যদি আমরা break না দিতাম
while True:
    num = int(input("একটি ঋণাত্মক সংখ্যা দিলে বন্ধ হবে: "))
    if num < 0:
        print("ধন্যবাদ, বন্ধ করা হলো!")
        break
```

### 📝 ক্লাস ০৮ — হোমওয়ার্ক
1. `while` লুপ ব্যবহার করে ১ থেকে ৫০ পর্যন্ত সংখ্যাগুলোর যোগফল বের করো।
2. ১ থেকে ২০ পর্যন্ত সংখ্যার লুপ চালাও এবং শুধুমাত্র বিজোড় সংখ্যাগুলো প্রিন্ট করো (`continue` ব্যবহার করে)।
