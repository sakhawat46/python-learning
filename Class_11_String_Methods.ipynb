### 🎯 আজকের লক্ষ্য
- স্ট্রিং এর জনপ্রিয় মেথডস (`upper()`, `lower()`, `replace()`, `split()`, `join()`, `find()`)
- Escape Characters (এস্কেপ ক্যারেক্টার) এর তালিকা ও ব্যবহার

```python
# ১. ক্যাস পরিবর্তন ও রিপ্লেস
txt = "  Welcome to Python Programming!  "
print("বড় হাতের:", txt.upper())
print("ছোট হাতের:", txt.lower())
print("স্পেস মুছে ফেলা (strip):", txt.strip())
print("প্রতিস্থাপন (replace):", txt.replace("Python", "AI"))
```

```python
# ২. ভাগ করা (split) এবং জোড়া লাগানো (join)
langs = "HTML,CSS,JavaScript,Python"
lang_list = langs.split(",")
print("স্প্লিট করা লিস্ট:", lang_list)

joined_txt = " - ".join(lang_list)
print("জয়েন করা স্ট্রিং:", joined_txt)
```

```python
# ৩. খোঁজা (find) ও চেক করা
mystr = "Learning is fun"
print("fun এর অবস্থান (index):", mystr.find("fun"))
print("is দিয়ে শেষ হয়েছে?", mystr.endswith("fun"))
```

```python
# ৪. Escape Characters
txt = "আমরা তথাকথিত \"ভাইকিংস\" উত্তর থেকে এসেছি।"
print(txt)
print("প্রথম লাইন।\nদ্বিতীয় লাইন (নতুন লাইন)")
print("বাম\tডান (ট্যাব স্পেস)")
```
