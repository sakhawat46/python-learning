# মডিউল ৫: Collections
## ক্লাস ১৪: Tuple & Set

### 🎯 আজকের লক্ষ্য
- Tuple তৈরি ও এর অপরিবর্তনশীলতা (Immutable) বোঝা
- Set তৈরি ও অনন্য মান (Unique values) রাখা
- সেটের বিভিন্ন অপারেশন (Union, Intersection, Difference)
- List, Tuple, এবং Set এর পার্থক্য জানা

```python
# ১. Tuple - ফার্স্ট ব্র্যাকেট (), এটি পরিবর্তন করা যায় না (Immutable)
days = ("রবি", "সোম", "মঙ্গল")
print("টিউপল:", days)
print("রবি কতবার আছে?", days.count("রবি"))

# Unpacking tuple with asterisks
fruits = ("apple", "banana", "cherry", "strawberry", "raspberry")
(green, yellow, *red) = fruits
print(green)
print(yellow)
print(red)
```

```python
# ২. Set - কার্লি ব্র্যাকেট {}, ডুপ্লিকেট মান বাদ যায় এবং এটি সাজানো থাকে না
set_a = {1, 2, 3, 3, 4}  # ৩ দুবার আছে
print("সেট a:", set_a)   # ডুপ্লিকেট বাদ যাবে

set_b = {3, 4, 5, 6}

# সেট অপারেশনসমূহ
print("সংযোগ (Union):", set_a.union(set_b))
print("ছেদ (Intersection):", set_a.intersection(set_b))
print("পার্থক্য (Difference):", set_a.difference(set_b))
```

### 📋 মূল পার্থক্যসমূহ:
| ফিচার | List | Tuple | Set |
| :--- | :--- | :--- | :--- |
| **ব্র্যাকেট** | Third Bracket `[]` | First Bracket `()` | Curly Bracket `{}` |
| **ডুপ্লিকেট মান** | অনুমতি দেয় | অনুমতি দেয় | অনুমতি দেয় না (Unique) |
| **অর্ডার (ক্রম)** | সাজানো (Ordered) | সাজানো (Ordered) | অগোছালো (Unordered) |
| **পরিবর্তনযোগ্যতা** | যায় (Mutable) | যায় না (Immutable) | যায় (আইটেম যোগ/বাদ দেওয়া যায়) |
