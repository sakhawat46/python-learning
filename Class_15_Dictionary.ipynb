# মডিউল ৫: Collections
## ক্লাস ১৫: Dictionary

### 🎯 আজকের লক্ষ্য
- Key-Value জোড় দিয়ে ডেটা রাখা
- ডিকশনারি আইটেম অ্যাক্সেস, আপডেট ও ডিলিট করা
- Nested Dictionary (ডিকশনারির ভেতরে ডিকশনারি)
- ফোনবুক ও বাস্তবমুখী প্রয়োগ

```python
# ১. ডিকশনারি ডিক্লেয়ারেশন ও অ্যাক্সেস
student = {
    "name": "সানা",
    "roll": 102,
    "gpa": 4.75,
    "college": "Dhaka College"
}

print("নাম:", student["name"])
print("কলেজ (get মেথড):", student.get("college"))
```

```python
# ২. আপডেট ও ডিলিট
student["gpa"] = 5.00                       # মান পরিবর্তন
student.update({"subject": "বিজ্ঞান"})      # নতুন কী যুক্ত
print("আপডেট করার পর:", student)

# ডিলিট
del student["college"]
popped_item = student.popitem()             # শেষের জোড়া রিমুভ করে
print("ডিলিট করার পর:", student)
print("বাদ দেওয়া আইটেম:", popped_item)
```

```python
# ৩. Nested Dictionary (ডিকশনারির ভেতরে ডিকশনারি)
my_family = {
    "child1": {"name": "রাফি", "age": 17},
    "child2": {"name": "সানা", "age": 15}
}
print("child1 এর নাম:", my_family["child1"]["name"])
```
