# মডিউল ২: Decision Making
## ক্লাস ৫: If-Elif-Else

### 🎯 আজকের লক্ষ্য
- একাধিক শর্তের সমন্বয়ে `if-elif-else` গঠন
- গ্রেড ক্যালকুলেটর (Grade Calculator) তৈরি
- বয়স ভেরিফিকেশন সিস্টেম (Age Verification System) তৈরি

```python
# ১. Grade Calculator (গ্রেড ক্যালকুলেটর)
marks = int(input("আইসিটি পরীক্ষার প্রাপ্ত নম্বর লিখো (০-১০০): "))

if marks >= 80:
    print("গ্রেড: A+ 🏆")
elif marks >= 70:
    print("গ্রেড: A")
elif marks >= 60:
    print("গ্রেড: A-")
elif marks >= 50:
    print("গ্রেড: B")
elif marks >= 40:
    print("গ্রেড: C")
elif marks >= 33:
    print("গ্রেড: D")
else:
    print("গ্রেড: F 😢")
```

```python
# ২. Age Verification System (টিকিট কাউন্টার লজিক)
age = int(input("বয়স লিখো: "))

if age < 5:
    print("টিকিট ফ্রি! 👶")
elif age <= 12:
    print("টিকিট মূল্য: ৫০ টাকা।")
elif age <= 60:
    print("টিকিট মূল্য: ১০০ টাকা।")
else:
    print("টিকিট মূল্য: ৭০ টাকা (বয়োজ্যেষ্ঠদের জন্য ছাড়)।")
```
