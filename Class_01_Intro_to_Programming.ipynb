### 1. Programming কী?
কম্পিউটারকে কোনো কাজ করানোর জন্য যে দিকনির্দেশনা বা কোড দেওয়া হয়, তাকে **Programming** বলে। যেমন মানুষের সাথে যোগাযোগের জন্য ভাষা লাগে, তেমনি কম্পিউটারের সাথে যোগাযোগের জন্য প্রোগ্রামিং ভাষা লাগে।

### 2. Software vs Program
- **Program**: নির্দিষ্ট কাজ করার জন্য কিছু কোড বা নির্দেশনার সমষ্টি।
- **Software**: একাধিক প্রোগ্রাম, ডকুমেন্টেশন ও কনফিগারেশন ফাইলের একটি পূর্ণাঙ্গ প্যাকেজ যা ইউজার সরাসরি ব্যবহার করতে পারে।

### 3. Python কেন শিখবো?
- পাইথনের সিনট্যাক্স খুব সহজ এবং ইংরেজি ভাষার মত।
- এটি ওয়েব ডেভেলপমেন্ট, কৃত্রিম বুদ্ধিমত্তা (AI), এবং ডেটা সায়েন্সে সবচেয়ে জনপ্রিয় ভাষা।
- HSC আইসিটি (ICT) পরীক্ষার প্রোগ্রামিং অংশে এটি শেখা খুবই উপযোগী।

### 4. এনভায়রনমেন্ট সেটআপ (Installation & VS Code)
- প্রথমে [python.org](https://www.python.org/) থেকে পাইথন ডাউনলোড করে ইন্সটল করুন (ইন্সটলেশনের সময় অবশ্যই **"Add Python to PATH"** বক্সে টিক মার্ক দিবেন)।
- এরপর কোড লেখার জন্য [Visual Studio Code](https://code.visualstudio.com/) ডাউনলোড করুন এবং এতে `Python` এক্সটেনশন ইন্সটল করুন।

```python
# প্রথম পাইথন প্রোগ্রাম!
print("Hello, World!")
print("আসসালামু আলাইকুম! আমি পাইথন শিখছি।")
```
