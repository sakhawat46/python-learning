## ক্লাস ২৩: Exception Handling

### 🎯 আজকের লক্ষ্য
- `try`, `except`, এবং `finally` ব্লকের ব্যবহার
- সাধারণ পাইথন ত্রুটিসমূহ (ValueError, ZeroDivisionError, etc.)
- কাস্টম এরর ছুড়ে মারা (`raise`)

```python
# ১. try-except দিয়ে এরর ধরা
try:
    number = int(input("একটি সংখ্যা দাও: "))
    result = 100 / number
    print("ফলাফল:", result)
except ValueError:
    print("❌ ত্রুটি: অনুগ্রহ করে কোনো ক্যারেক্টার না দিয়ে সংখ্যা দিন।")
except ZeroDivisionError:
    print("❌ ত্রুটি: শূন্য (০) দিয়ে ভাগ করা অসম্ভব।")
except Exception as e:
    print("❌ অন্য কোনো ত্রুটি:", e)
finally:
    print("এটি সবসময় চলবে (Cleanup block)")
```

```python
# ২. কাস্টম এরর তৈরি (raise)
def check_age(age):
    if age < 0:
        raise ValueError("বয়স ঋণাত্মক হতে পারে না!")
    return f"বয়স {age} গ্রহণ করা হলো।"

try:
    print(check_age(-5))
except ValueError as err:
    print("ধরা পড়া এরর:", err)
```
