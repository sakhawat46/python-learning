## ক্লাস ১৮: Advanced Functions

### 🎯 আজকের লক্ষ্য
- Default Arguments ও Keyword Arguments
- Variable Length Arguments (`*args` এবং `**kwargs`)
- Local vs Global Scope এবং `global` কিওয়ার্ড

```python
# ১. Default ও Keyword Arguments
def power(base, exp=2):  # exp এর ডিফল্ট মান ২
    return base ** exp

print("ডিফল্ট ঘাত (৫²):", power(5))
print("নির্দিষ্ট ঘাত (৫³):", power(5, 3))

# Keyword arguments
print(power(exp=4, base=2))  # ২⁴ = ১৬
```

```python
# ২. Variable Length Arguments (*args, **kwargs)
def sum_all(*numbers):  # *args (টিউপল হিসেবে নেয়)
    return sum(numbers)

print("যোগফল:", sum_all(1, 2, 3, 4, 5))

def print_details(**info):  # **kwargs (ডিকশনারি হিসেবে নেয়)
    for key, val in info.items():
        print(f"{key}: {val}")

print_details(name="করিম", roll=101, college="Dhaka College")
```

```python
# ৩. Scope (লোকাল বনাম গ্লোবাল)
x = 100  # গ্লোবাল

def change_global():
    global x  # গ্লোবাল ভ্যারিয়েবল মডিফাই করতে
    x = 200

change_global()
print("পরিবর্তিত গ্লোবাল x:", x)
```
