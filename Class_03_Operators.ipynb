## ক্লাস ৩: Operators

### 🎯 আজকের লক্ষ্য
- Arithmetic, Comparison, Assignment, Logical অপারেটরস
- Identity (`is`) এবং Membership (`in`) অপারেটরস
- Operator Precedence (অগ্রাধিকার ক্রম) এবং প্র্যাকটিস

```python
# Arithmetic (গাণিতিক) অপারেটর
a = 15
b = 4

print("যোগ:", a + b)
print("বিয়োগ:", a - b)
print("গুণ:", a * b)
print("ভাগ:", a / b)
print("পূর্ণ ভাগ (Floor division):", a // b)
print("ভাগশেষ (Modulus):", a % b)
print("ঘাত (Exponentiation):", a ** b)
```

```python
# Comparison (তুলনা) এবং Logical (যুক্তি) অপারেটর
x = 10
y = 20

print("সমান?", x == y)
print("ছোট?", x < y)
print("সমান নয়?", x != y)

# Logical AND, OR, NOT
print(x < 15 and y > 15)  # দুটিই সত্য হলে True
print(x > 15 or y > 15)   # যেকোনো একটি সত্য হলে True
print(not(x == y))         # উল্টো রেজাল্ট দেবে
```

```python
# Identity & Membership Operators
names = ["সাকিব", "তামিম"]
print("সাকিব" in names)      # True (Membership)
print("মুশফিক" not in names)  # True

list1 = [1, 2]
list2 = [1, 2]
list3 = list1
print(list1 is list3)       # True (Identity - একই মেমোরি)
print(list1 is list2)       # False (আলাদা মেমোরি)
```
