# মডিউল ৮: Object-Oriented Programming (OOP)
## ক্লাস ২৭: Polymorphism, Dunder Methods & Abstraction

### 🎯 আজকের লক্ষ্য
- Polymorphism (বহুরূপতা) এবং Duck Typing এর ধারণা
- Magic / Dunder Methods: `__str__`, `__repr__`, `__len__`
- Operator Overloading: `__add__` ও `__eq__` দিয়ে অবজেক্ট অপারেশন
- Abstraction এবং Abstract Base Classes (`abc.ABC`, `@abstractmethod`)

```python
# ১. Polymorphism ও Duck Typing (একই ইন্টারফেস কিন্তু ভিন্ন ভিন্ন আচরণ)

class Cat:
    def sound(self):
        return "মিউ মিউ 🐱"

class Dog:
    def sound(self):
        return "ঘেউ ঘেউ 🐶"

class Cow:
    def sound(self):
        return "হাম্বা হাম্বা 🐮"

# সাধারণ ফাংশন যা যেকোনো প্রাণীর অবজেক্ট গ্রহণ করতে পারে
def make_sound(animal):
    # Duck Typing: "যদি এটি হাঁসের মতো ডাকে তবে এটি হাঁস"
    print("ডাক:", animal.sound())

animals = [Cat(), Dog(), Cow()]
for a in animals:
    make_sound(a)
```

```python
# ২. Magic / Dunder Methods এবং Operator Overloading (অপারেটর ওভারলোডিং)
# নামের আগে ও পরে ডাবল আন্ডারস্কোর (__মেথড__) থাকে বলেই এদের Dunder Method বলে।

class Book:
    def __init__(self, title, author, pages):
        self.title = title
        self.author = author
        self.pages = pages

    # print() বা str() কল করলে ইউজার-ফ্রেন্ডলি আউটপুট
    def __str__(self):
        return f"'{self.title}' — {self.author} ({self.pages} পৃষ্ঠা)"

    # ইন্টারপ্রেটার ও ডিবাগিংয়ের জন্য বিস্তারিত রূপ
    def __repr__(self):
        return f"Book('{self.title}', '{self.author}', {self.pages})"

    # len(book) কল করলে পৃষ্ঠার সংখ্যা প্রদান
    def __len__(self):
        return self.pages

    # '+' অপারেটর ওভারলোডিং: দুটি বইয়ের মোট পৃষ্ঠা যোগ
    def __add__(self, other):
        return self.pages + other.pages

    # '==' অপারেটর ওভারলোডিং: দুটি বই একই কি না যাচাই
    def __eq__(self, other):
        return self.title == other.title and self.author == other.author

b1 = Book("পাইথন পরিচিতি", "সাখাওয়াত", 250)
b2 = Book("অ্যাডভান্সড পাইথন", "সাখাওয়াত", 350)
b3 = Book("পাইথন পরিচিতি", "সাখাওয়াত", 250)

print(b1)                           # __str__ কল হবে
print(repr(b2))                     # __repr__ কল হবে
print("বইটির মোট পৃষ্ঠা:", len(b1))       # __len__ কল হবে
print("দুটি বইয়ের মোট পৃষ্ঠা যোগফল (+):", b1 + b2) # __add__ কল হবে
print("b1 ও b3 কি একই বই (==)?:", b1 == b3)        # __eq__ কল হবে
```

```python
# ৩. Abstraction ও Abstract Base Class (ABC)
# অ্যাবস্ট্রাক্ট ক্লাস একটি ব্লুপ্রিন্ট (কঙ্কাল) তৈরি করে, যার মেথডগুলো চাইল্ড ক্লাসে বাস্তবায়ন বাধ্যতামূলক।

from abc import ABC, abstractmethod

class PaymentGateway(ABC):
    @abstractmethod
    def pay(self, amount):
        pass  # চাইল্ড ক্লাসে অবশ্যই থাকতে হবে

    @abstractmethod
    def refund(self, amount):
        pass

class BkashPayment(PaymentGateway):
    def pay(self, amount):
        print(f"বিকাশের মাধ্যমে {amount} টাকা পেমেন্ট সফল হয়েছে! 📱")

    def refund(self, amount):
        print(f"বিকাশ অ্যাকাউন্টে {amount} টাকা রিফান্ড পাঠানো হয়েছে।")

class NagadPayment(PaymentGateway):
    def pay(self, amount):
        print(f"নগদের মাধ্যমে {amount} টাকা পেমেন্ট সম্পন্ন হয়েছে! 💸")

    def refund(self, amount):
        print(f"নগদ অ্যাকাউন্টে {amount} টাকা ফেরত এসেছে।")

# gateway = PaymentGateway() # এটি TypeError দিবে কারণ অ্যাবস্ট্রাক্ট ক্লাসের সরাসরি অবজেক্ট তৈরি হয় না

payment1 = BkashPayment()
payment1.pay(1200)

payment2 = NagadPayment()
payment2.pay(2500)
payment2.refund(500)
```
