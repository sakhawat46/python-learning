# মডিউল ৬: Functions
## ক্লাস ১৬: Functions Basics

### 🎯 আজকের লক্ষ্য
- Function এর সংজ্ঞা ও প্রয়োজনীয়তা
- Parameter বনাম Argument এর পার্থক্য বোঝা
- Return স্টেটমেন্টের মাধ্যমে মান ফেরত পাঠানো

```python
# ১. সাধারণ ফাংশন তৈরি (def দিয়ে শুরু)
def greet():
    print("হ্যালো! পাইথন ফাংশন ক্লাসে স্বাগতম।")

greet()  # ফাংশন কল
```

```python
# ২. Parameter ও Argument সহ ফাংশন
# name এখানে Parameter (চলক)
def greet_student(name):
    print(f"আসসালামু আলাইকুম, {name}!")

# "রাফি" এখানে Argument (প্রকৃত মান)
greet_student("রাফি")
```

```python
# ৩. Return স্টেটমেন্ট সহ ফাংশন
def calculate_square(num):
    return num * num

result = calculate_square(5)
print("৫ এর বর্গ:", result)
```

2. একটি সংখ্যা জোড় নাকি বিজোড় তা চেক করার ফাংশন `is_even(n)` তৈরি করো যা True বা False রিটার্ন করবে।
