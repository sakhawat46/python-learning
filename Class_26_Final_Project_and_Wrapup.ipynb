### 🎯 আজকের লক্ষ্য
- পূর্ণাঙ্গ স্টুডেন্ট রেজাল্ট ম্যানেজমেন্ট সিস্টেম (Student Result Management System) কোড ডেমো
- কোডিং ডিবাগিং ও প্র্যাকটিক্যাল গাইড
- ভাইভা (Viva) সেশন প্রস্তুতি
- পুরো পাইথন কোর্সের ফাইনাল রিভিউ

```python
# ১. Student Result Management System (পূর্ণাঙ্গ অবজেক্ট অরিয়েন্টেড প্রজেক্ট)
import json

class Student:
    def __init__(self, id, name, marks_dict):
        self.id = id
        self.name = name
        self.marks = marks_dict
        
    def get_average(self):
        return sum(self.marks.values()) / len(self.marks)
        
    def get_gpa(self):
        avg = self.get_average()
        if avg >= 80: return 5.00
        elif avg >= 70: return 4.00
        elif avg >= 60: return 3.50
        elif avg >= 50: return 3.00
        elif avg >= 33: return 2.00
        else: return 0.00
        
    def show_info(self):
        print(f"\nID: {self.id} | নাম: {self.name}")
        print("নম্বরসমূহ:")
        for sub, mark in self.marks.items():
            print(f"  {sub}: {mark}")
        print(f"  গড় নম্বর: {self.get_average():.2f} | GPA: {self.get_gpa():.2f}")

# ডেমো ডেটা দিয়ে রান করা
marks = {"বাংলা": 82, "ইংরেজি": 75, "গণিত": 90}
student1 = Student(101, "রাফি আহমেদ", marks)
student1.show_info()
```
