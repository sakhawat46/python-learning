## ক্লাস ২০: Closures, Decorators & Generators

### 🎯 আজকের লক্ষ্য
- First-Class Functions ও Closures এর ধারণা
- Python Decorator কী এবং `@decorator` সিনট্যাক্সের ব্যবহার
- `*args` ও `**kwargs` সহ রিয়্যূজেবল ডেকোরেটর তৈরি
- প্র্যাকটিক্যাল এক্সিকিউশন টাইম (Timer) ডেকোরেটর তৈরি
- Python Generators ও `yield` স্টেটমেন্ট দিয়ে মেমরি এফিশিয়েন্ট কোড লেখা

```python
# ১. First-Class Functions এবং Closures
# পাইথনে ফাংশন একটি অবজেক্ট (First-Class Citizen), তাই একে চলকে রাখা বা আর্গুমেন্ট হিসেবে পাঠানো যায়।

# Closure: ভেতরের ফাংশন বাইরের স্কোপের ভ্যারিয়েবল মনে রাখে
def multiplier(factor):
    def multiply(number):
        return number * factor
    return multiply

double = multiplier(2)
triple = multiplier(3)

print("১০ এর দ্বিগুণ:", double(10))
print("১০ এর তিনগুণ:", triple(10))
```

```python
# ২. Python Decorator (ফাংশনের ভেতরে পরিবর্তন না করে অতিরিক্ত ফিচার যোগ করা)
# একটি ডেকোরেটর ফাংশন অন্য ফাংশনকে র‍্যাপ (wrap) করে তার আচরণ সম্প্রসারিত করে।

def my_logger(func):
    def wrapper(*args, **kwargs):
        print(f"[LOG]: '{func.__name__}' ফাংশন কল করা হচ্ছে...")
        result = func(*args, **kwargs)
        print(f"[LOG]: '{func.__name__}' সম্পন্ন হয়েছে!")
        return result
    return wrapper

# @ সিনট্যাক্স দিয়ে ডেকোরেটরের প্রয়োগ:
@my_logger
def greet(name):
    print(f"হ্যালো, {name}!")

greet("সাকিব")
```

```python
# ৩. বাস্তবমুখী টাইমার ডেকোরেটর (Execution Time Measurement)
import time

def execution_timer(func):
    def wrapper(*args, **kwargs):
        start_time = time.time()
        result = func(*args, **kwargs)
        end_time = time.time()
        elapsed = end_time - start_time
        print(f"⏱️ '{func.__name__}' সম্পন্ন হতে সময় লেগেছে: {elapsed:.6f} সেকেন্ড")
        return result
    return wrapper

@execution_timer
def calculate_big_sum():
    return sum(range(1, 1000001))

total = calculate_big_sum()
print("১ থেকে ১০ লক্ষের যোগফল:", total)
```

```python
# ৪. Python Generators ও `yield` কিওয়ার্ড (Memory Efficiency)
# সাধারণ ফাংশনে return করলে পুরো ডেটা একবারে মেমরিতে জমা হয়।
# yield ব্যবহারকারী জেনারেটর অন-ডিমান্ড (On-Demand) একেকটি মান তৈরি করে মেমরি বাঁচায়।

def countdown_generator(n):
    print("কাউন্টডাউন শুরু...")
    while n > 0:
        yield n  # পজিশন মনে রেখে মান পাঠায় এবং পজ করে
        n -= 1

gen = countdown_generator(3)
print("প্রথম কল:", next(gen))
print("দ্বিতীয় কল:", next(gen))
print("তৃতীয় কল:", next(gen))

# লুপের সাহায্যে জেনারেটর চালানো (ফিবোনাচ্চি জেনারেটর)
def fibonacci_gen(limit):
    a, b = 0, 1
    for _ in range(limit):
        yield a
        a, b = b, a + b

print("\nফিবোনাচ্চি জেনারেটরের প্রথম ৭টি সংখ্যা:")
for num in fibonacci_gen(7):
    print(num, end=" ")
print()

# জেনারেটর এক্সপ্রেশন (Generator Expression)
square_gen = (x * x for x in range(1, 6))
print("Generator Expression থেকে লিস্ট:", list(square_gen))
```
