### 🎯 আজকের লক্ষ্য
- Inheritance (উত্তরাধিকার) ও `super()` ফাংশনের ব্যবহার
- Method Overriding (মেথড ওভাররাইডিং)
- Encapsulation (এনক্যাপসুলেশন - অ্যাক্সেস রেস্ট্রিকশন)

```python
# ১. Inheritance এবং super()
class Person:
    def __init__(self, name, age):
        self.name = name
        self.age = age
    
    def introduce(self):
        print(f"আমার নাম {self.name}। বয়স {self.age} বছর।")

class Teacher(Person):
    def __init__(self, name, age, subject):
        super().__init__(name, age)  # প্যারেন্ট ক্লাসের কনস্ট্রাকটর কল
        self.subject = subject
    
    # ২. Method Overriding (প্যারেন্ট ক্লাসের মেথড নিজের মত সাজানো)
    def introduce(self):
        print(f"আসসালামু আলাইকুম, আমি {self.name} স্যার। আমি {self.subject} পড়াই।")

teacher = Teacher("সাখাওয়াত", 30, "Python")
teacher.introduce()
```

```python
# ৩. Encapsulation (তথ্য গোপন রাখা)
# পাইথনে ভ্যারিয়েবলের নামের শুরুতে ডাবল আন্ডারস্কোর (__) দিলে তা প্রাইভেট হয়ে যায়
class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self.__balance = balance  # প্রাইভেট এট্রিবিউট
        
    # ব্যালেন্স দেখার নিরাপদ মেথড (Getter)
    def get_balance(self):
        return self.__balance
        
    # ব্যালেন্স জমা করার মেথড (Setter)
    def deposit(self, amount):
        if amount > 0:
            self.__balance += amount
            print(f"{amount} টাকা জমা হয়েছে।")
        else:
            print("অবৈধ পরিমাণ!")

acc = BankAccount("রাফি", 5000)
# print(acc.__balance)  # এটি সরাসরি অ্যাক্সেস করতে দিবে না (Error আসবে)
print("বর্তমান ব্যালেন্স:", acc.get_balance())
acc.deposit(1500)
print("নতুন ব্যালেন্স:", acc.get_balance())
```
