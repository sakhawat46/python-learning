### ১. Reverse a String (স্ট্রিং উল্টানো)
**Main Logic:** প্রতিটা character কে string এর **শুরুতে** বসাতে হবে (prepend), তাহলে শেষের character টা সবার আগে চলে আসবে। এভাবে string reverse হয়ে যায়।
- `result = i + result` → নতুন character কে আগের result এর আগে জোড়া লাগানো হচ্ছে।

```python
# ক) পাইথন স্লাইসিং মেথড (সহজ উপায়)
name = "Sami"
print("Slicing ব্যবহার করে:", name[::-1])

# খ) লুপ ব্যবহার করে রিভার্স করা
word = "Sami"
result = ""
for i in word:
    result = i + result

print("লুপ ব্যবহার করে:", result)
```

### ২. Check Palindrome (প্যালিনড্রোম যাচাই)
**Main Logic:** string টাকে reverse করে original string এর সাথে compare করলেই বোঝা যাবে এটা palindrome কি না।
- `result == word` হলে palindrome, নাহলে না।

```python
word = "madam"
result = ""
for i in word:
    result = i + result

if result == word:
    print(f"'{word}' একটি Palindrome")
else:
    print(f"'{word}' Palindrome নয়")
```

### ৩. Find Duplicate Elements (ডুপ্লিকেট উপাদান খুঁজে বের করা)
**Main Logic:** Nested loop দিয়ে প্রতিটা element কে তার পরের সব element এর সাথে compare করা হচ্ছে। মিল পেলে এবং আগে থেকে duplicates লিস্টে না থাকলে add করা হচ্ছে।
- Time complexity: O(n²)

```python
nums = [1, 2, 3, 4, 2, 5, 1, 1]

duplicates = []
n = len(nums)
for i in range(n):
    for j in range(i + 1, n):
        if nums[i] == nums[j] and nums[i] not in duplicates:
            duplicates.append(nums[i])

print("ডুপ্লিকেট উপাদানগুলো:", duplicates)
```

### ৪. Count Character Frequency (বর্ণের উপস্থিতি গণনা)
**Main Logic:** dictionary ব্যবহার করে প্রতিটা character কে key হিসেবে রেখে তার count বাড়ানো হচ্ছে।
- Character আগে থেকে থাকলে count +1, না থাকলে নতুন key তৈরি করে 1 বসানো হচ্ছে।

```python
text = "sakhawat"
freq = {}
for i in text:
    if i in freq:
        freq[i] = freq[i] + 1
    else:
        freq[i] = 1

print("ক্যারেক্টার ফ্রিকোয়েন্সি:", freq)
```

### ৫. Linear Search (লিনিয়ার সার্চ)
**Main Logic:** List এর শুরু থেকে শেষ পর্যন্ত একটা একটা করে element চেক করা হচ্ছে। Target পেলে তার index রাখা হচ্ছে, না পেলে -1 থেকে যাবে।
- Time complexity: O(n)

```python
nums = [10, 50, 30, 70, 80, 20]
target = 30

index = -1
for i in range(len(nums)):
    if nums[i] == target:
        index = i
        break

print(f"উপাদান {target} এর ইনডেক্স:", index)
```

### ৬. Bubble Sort (বাবল সর্ট)
**Main Logic:** পাশাপাশি দুইটা element compare করে বড়টা যদি আগে থাকে তাহলে swap করা হয়। প্রতি pass এর শেষে সবচেয়ে বড় element শেষে চলে যায়।
- Outer loop: কতবার pass চালাতে হবে।
- Inner loop: প্রতি pass এ কতগুলো comparison লাগবে।
- Time complexity: O(n²)

```python
nums = [64, 34, 25, 12, 22, 11, 90]

n = len(nums)
for i in range(n):
    for j in range(0, n - i - 1):
        if nums[j] > nums[j + 1]:
            nums[j], nums[j + 1] = nums[j + 1], nums[j]

print("বাবল সর্টের পর সাজানো লিস্ট:", nums)
```

### ৭. FizzBuzz
**Main Logic:** কোন সংখ্যা ৩ ও ৫ দুটো দিয়েই ভাগ যায় কি না সেটা আগে চেক করতে হবে (অর্থাৎ ১৫ দিয়ে ভাগ যায় কি না), কারণ এটাই সবচেয়ে specific condition। এরপর আলাদাভাবে ৩ ও ৫ চেক করা হয়।
- **Order matters:** ১৫ এর check সবার আগে না করলে ভুল output আসবে।

```python
print("১ থেকে ২০ এর জন্য FizzBuzz ফলাফল:")
for i in range(1, 21):          # ১ থেকে ২০ পর্যন্ত লুপ চলবে
    if i % 15 == 0:             # ১. প্রথমে ১৫ দিয়ে চেক করা হয়েছে
        print("FizzBuzz")
    elif i % 3 == 0:            # ২. শুধু ৩ দিয়ে ভাগ যায় কি না
        print("Fizz")
    elif i % 5 == 0:            # ৩. শুধু ৫ দিয়ে ভাগ যায় কি না
        print("Buzz")
    else:                       # ৪. কোনো শর্ত পূরণ না হলে সংখ্যাটি বসবে
        print(i)
```

### ৮. Check Prime Number (মৌলিক সংখ্যা নির্ণয়)
**Main Logic:** ২ থেকে √n পর্যন্ত সংখ্যাগুলো দিয়ে ভাগ করে দেখা হয়, কোনোটা দিয়ে নিঃশেষে ভাগ যায় কি না। ভাগ গেলে prime না।
- **কেন √n পর্যন্ত চেক করলেই যথেষ্ট:** কোনো factor √n এর বেশি হলে, তার জোড়া factor টা অবশ্যই √n এর কম হবে — তাই সেটা আগেই ধরা পড়ে যাবে।
- ২ এর কম সংখ্যা prime না — সেটা আগেই বাদ দেওয়া হয়েছে।

```python
n = 17
is_prime = True

if n < 2:
    is_prime = False
else:
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            is_prime = False
            break

print(f"{n} মৌলিক সংখ্যা কি না?:", is_prime)
```

### ৯. Fibonacci Series (ফিবোনাচ্চি ধারা)
**Main Logic:** দুইটা variable (`a`, `b`) দিয়ে আগের দুইটা সংখ্যার যোগফল রাখা হয়। প্রতি loop এ `a` প্রিন্ট হয়, তারপর `a, b = b, a+b` করে দুটো variable এগিয়ে যায়।
- Space efficient: পুরো list store না করে শুধু last দুইটা number রেখেই series generate করা হচ্ছে।

```python
n = 10
a, b = 0, 1

print(f"ফিবোনাচ্চি সিরিজের প্রথম {n}টি পদ:")
for _ in range(n):
    print(a, end=" ")
    a, b = b, a + b
print()
```

### ১০. Check Leap Year (লিপ ইয়ার যাচাই)
**Main Logic:** একটা বছর leap year হবে যদি —
1. ৪০০ দিয়ে নিঃশেষে ভাগ যায় (যেমন 2000), **অথবা**
2. ৪ দিয়ে ভাগ যায় কিন্তু ১০০ দিয়ে ভাগ যায় না (যেমন 2024)।

- এই দুই condition এর মধ্যে যেকোনো একটা সত্যি হলেই leap year (`or` ব্যবহার করা হয়েছে)।

```python
year = 2024

if (year % 400 == 0) or (year % 4 == 0 and year % 100 != 0):
    print(f"{year} is a Leap Year")
else:
    print(f"{year} is NOT a Leap Year")
```
