### 🎯 আজকের লক্ষ্য
- পাইথনে String তৈরি করার বিভিন্ন উপায়
- পজিটিভ ও নেগেটিভ Indexing (সূচক)
- String Slicing (স্ট্রিং কেটে নেওয়া)

```python
# ১. স্ট্রিং তৈরি ও মাল্টিলাইন স্ট্রিং
s1 = 'Hello'
s2 = "Python Programming"
multiline = """এটি একটি বহু-লাইনের স্ট্রিং।
এখানে নতুন লাইন ব্যবহার করা যায় সহজে।"""

print(multiline)
print("দৈর্ঘ্য:", len(s2))
```

```python
# ২. Indexing (নির্দিষ্ট ক্যারেক্টার দেখা)
text = "BANGLADESH"
print("প্রথম ক্যারেক্টার (Index 0):", text[0])
print("তৃতীয় ক্যারেক্টার (Index 2):", text[2])
print("শেষ ক্যারেক্টার (Index -1):", text[-1])
print("শেষের আগের ক্যারেক্টার (Index -2):", text[-2])
```

```python
# ৩. Slicing (স্ট্রিং কাটা) -> text[start:end:step]
print("Slice (0 to 5):", text[0:6])
print("Slice (from start to 6):", text[:6])
print("Slice (from index 6 to end):", text[6:])
print("ধাপসহ স্লাইস (০ থেকে ১০ প্রতি ২ ধাপে):", text[0:10:2])
print("স্ট্রিং উল্টানো (Reverse):", text[::-1])
```
