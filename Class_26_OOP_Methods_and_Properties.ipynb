# মডিউল ৮: Object-Oriented Programming (OOP)
## ক্লাস ২৬: Class Methods, Static Methods & Property Decorator

### 🎯 আজকের লক্ষ্য
- Class Variables বনাম Instance Variables এর পার্থক্য বোঝা
- Instance Method, Class Method (`@classmethod`) ও Static Method (`@staticmethod`)
- Class Method দিয়ে অল্টারনেটিভ কনস্ট্রাকটর (Alternative Constructor) তৈরি
- Static Method দিয়ে স্বাধীন হেল্পার ইউটিলিটি তৈরি
- `@property`, `@setter` এবং `@deleter` দিয়ে আধুনিক পাইথনিক এনক্যাপসুলেশন

```python
# ১. Class Variable বনাম Instance Variable
# Class Variable ক্লাসের সকল অবজেক্টের জন্য একটাই মেমরি শেয়ার করে।
# Instance Variable প্রতিটি অবজেক্টের নিজস্ব এবং আলাদা।

class Student:
    school_name = "আইডিয়াল স্কুল অ্যান্ড কলেজ"  # ক্লাস ভ্যারিয়েবল (সকলের জন্য এক)
    total_students = 0                          # মোট শিক্ষার্থী সংখ্যা ট্র্যাকিং

    def __init__(self, name, roll):
        self.name = name                        # ইনস্ট্যান্স ভ্যারিয়েবল
        self.roll = roll                        # ইনস্ট্যান্স ভ্যারিয়েবল
        Student.total_students += 1             # নতুন অবজেক্ট তৈরির সাথে সাথে বৃদ্ধি পাবে

s1 = Student("রাফি", 101)
s2 = Student("সাদিয়া", 102)

print(f"{s1.name}-এর স্কুল: {s1.school_name}")
print(f"{s2.name}-এর স্কুল: {s2.school_name}")
print("মোট রেজিস্টার্ড ছাত্র-ছাত্রী:", Student.total_students)
```

```python
# ২. Instance Method, Class Method (@classmethod) ও Static Method (@staticmethod)

class Employee:
    company = "TechBD Solutions"
    base_salary = 30000

    def __init__(self, name, salary):
        self.name = name
        self.salary = salary

    # ক) Instance Method: 'self' এর মাধ্যমে অবজেক্টের নিজস্ব ডেটা নিয়ে কাজ করে
    def display_details(self):
        print(f"কর্মী: {self.name} | বেতন: {self.salary} টাকা | কোম্পানি: {self.company}")

    # খ) Class Method: 'cls' গ্রহণ করে এবং ক্লাস-লেভেল স্টেট বা ফ্যাক্টরি তৈরিতে লাগে
    @classmethod
    def from_string(cls, emp_str):
        # যেমন: "করিম-45000" থেকে সরাসরি অবজেক্ট তৈরি
        name, salary = emp_str.split("-")
        return cls(name, int(salary))

    # গ) Static Method: ক্লাসের কোনো স্টেট (self বা cls) লাগে না, সাধারণ লজিক্যাল ইউটিলিটি
    @staticmethod
    def is_workday(day):
        # শুক্র ও শনি ছুটির দিন ধরে
        weekend = ["Friday", "Saturday"]
        return day not in weekend

# সাধারণ কনস্ট্রাকটর দিয়ে অবজেক্ট তৈরি
emp1 = Employee("রাকিব", 35000)
emp1.display_details()

# Class Method (অল্টারনেটিভ কনস্ট্রাকটর) দিয়ে অবজেক্ট তৈরি
emp2 = Employee.from_string("তানভীর-50000")
emp2.display_details()

# Static Method কল করা
print("রবিবার কি কাজের দিন?:", Employee.is_workday("Sunday"))
print("শুক্রবার কি কাজের দিন?:", Employee.is_workday("Friday"))
```

```python
# ৩. Property Decorators (@property, @setter, @deleter)
# পাইথনিক এনক্যাপসুলেশন: মেথডকে সাধারণ ভ্যারিয়েবলের মতো এক্সেস করা ও ভ্যালিডেশন যোগ করা

class Product:
    def __init__(self, name, price):
        self.name = name
        self._price = price  # সুরক্ষিত ভ্যারিয়েবল

    # Getter প্রোপার্টি: product.price লিখলেই এটি কল হবে
    @property
    def price(self):
        return self._price

    # Setter প্রোপার্টি: product.price = 500 লিখলে ভ্যালিডেশন চেক হবে
    @price.setter
    def price(self, new_price):
        if new_price < 0:
            raise ValueError("দাম কখনো ঋণাত্মক হতে পারে না!")
        self._price = new_price
        print(f"দাম সফলভাবে আপডেট হয়েছে: {self._price} টাকা")

    # Deleter প্রোপার্টি: del product.price লিখলে কল হবে
    @price.deleter
    def price(self):
        print("দামের তথ্য রিসেট করা হলো।")
        self._price = 0

laptop = Product("Laptop", 75000)
print(f"{laptop.name} এর বর্তমান দাম:", laptop.price)

# সেটার কল হচ্ছে
laptop.price = 72000

# ডিলিটার কল হচ্ছে
del laptop.price
print("রিসেটের পর দাম:", laptop.price)
```
